```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
```

```python
X = np.array([[1, 7], [2, 8], [3, 8],  
              [5, 1], [6, -1], [7, 3]]) 
```

```python
y = [0, 0, 0, 1, 1, 1]
```

```python
clf = svm.SVC(kernel='linear') 
clf.fit(X, y)
```

```python
new_point = [[3, 5]]
prediction = clf.predict(new_point)

print(f"Prediction for {new_point}: Class {prediction[0]}")
```

```python
plt.scatter(X[np.array(y)==0][:, 0], X[np.array(y)==0][:, 1], color='blue', s=100, label='Class 0')

plt.scatter(X[np.array(y)==1][:, 0], X[np.array(y)==1][:, 1], color='brown', s=100, label='Class 1')

plt.scatter(new_point[0][0], new_point[0][1], c='green', marker='x', s=100, label='New Point')
```

```python
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

xx = np.linspace(xlim[0], xlim[1], 30)
yy = np.linspace(ylim[0], ylim[1], 30)
YY, XX = np.meshgrid(yy, xx)
xy = np.vstack([XX.ravel(), YY.ravel()]).T
Z = clf.decision_function(xy).reshape(XX.shape)

# Plot the lines
ax.contour(XX, YY, Z, colors='k', levels=[-1, 0, 1], alpha=0.5, linestyles=['--', '-', '--'])

# 5. Now plt.legend() will work because we added labels above
plt.legend(loc='upper left')
plt.show()
```
