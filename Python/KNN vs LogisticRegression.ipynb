```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
X = [[2, 50], [1, 30], [8, 90], [9, 95], [3, 40], [7, 85]] 
y = [0,       0,       1,       1,       0,       1]
```

sample dataset where 0 indicate fail and 1 indicate pass

```python
X_test = [[32.12], [89,56]]
y_test = [0, 1]
```

Logistic regression accuracy finding

```python
log_reg = LogisticRegression()
log_reg.fit(X, y)
log_acc = accuracy_score(y_test, log_reg.predict(X_test))
print(f"Logistic Regression Accuracy: {log_acc * 100}%")
```

KNN accuracy finding

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)
knn_acc = accuracy_score(y_test, knn.predict(X_test))
print(f"KNN accuracy: {knn_acc*100}%")
```

KNN and logistic regression accuracy is calculated and based on the value we can select the best model for that particular dataset.
