```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('SeoulBikeData.csv', encoding='unicode_escape')
```

```python
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df['Day'] = df['Date'].dt.day
```

```python
df.columns = ['Date', 'Rented_Bike_Count', 'Hour', 'Temperature', 'Humidity', 
              'Wind_speed', 'Visibility', 'Dew_point_temp', 'Solar_Radiation', 
              'Rainfall', 'Snowfall', 'Seasons', 'Holiday', 'Functioning_Day']

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Rented_Bike_Count'], kde=True, color='blue')
plt.title('Distribution of Rented Bike Counts')
plt.xlabel('Rented Bike Count')
plt.ylabel('Frequency')
plt.show()
```

Distribution of Bike Rentals
Here, I visualized how the number of rented bikes is distributed. The histogram shows that most of the time, the rental count is on the lower side, but there are some instances where the demand is very high. The curve (KDE) helps us see the overall shape of this distribution.

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='Hour', y='Rented_Bike_Count', data=df, palette='viridis')
plt.title('Average Rented Bike Count by Hour')
plt.show()
```

Average Bike Rentals by Hour
I wanted to see which time of the day is busiest. The bar chart clearly shows two major peaks: one in the morning around 8 AM and another in the evening around 6 PM. This suggests that a lot of people use these bikes for their daily commute to work or school.

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Seasons', y='Rented_Bike_Count', data=df, palette='magma')
plt.title('Average Bike Rentals per Season')
plt.show()
```

Seasonal Impact on Rentals
This plot compares the average number of bike rentals across different seasons. As expected, "Summer" has the highest usage because the weather is nice. Winter has the lowest rentals, likely due to the cold weather making it difficult to ride.

```python
sns.scatterplot(x='Temperature', y='Rented_Bike_Count', data=df, alpha=0.5)
plt.title('Temperature vs. Rented Bike Count')
plt.show()
```

Relationship Between Temperature and Rentals
I used a scatter plot to see how temperature affects bike usage. There is a clear trend: as the temperature rises, the number of rented bikes increases. However, once it gets too hot (around 35°C), the usage seems to stabilize or drop slightly.

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

Correlation Matrix
To understand which factors are most related to bike rentals, I created a heatmap. It shows that **Temperature** and **Hour** have the strongest positive correlation with bike counts. On the other hand, **Humidity** seems to have a negative effect—people ride less when it's very humid.

```python
sns.pointplot(x='Hour', y='Rented_Bike_Count', hue='Seasons', data=df)
plt.title('Hourly Bike Usage by Season')
plt.show()
```

Hourly Trends Across Seasons
This plot breaks down the hourly usage by season. Interestingly, the morning and evening rush hours are visible in every season. However, in **Winter**, the overall volume is much lower, and the morning peak is less sharp compared to the other seasons.

```python
sns.boxplot(x='Holiday', y='Rented_Bike_Count', data=df, palette='Set2')
plt.title('Bike Rentals: Holiday vs. No Holiday')
plt.show()
```

Holiday vs. Non-Holiday Rentals
I compared bike rentals on holidays versus regular days. Surprisingly, the median usage is quite similar, but the spread of data shows that non-holidays (working days) have more extreme outliers, probably due to the high volume of commuters.

```python
sns.barplot(x='Functioning_Day', y='Rented_Bike_Count', data=df)
plt.title('Rentals on Functioning vs. Non-Functioning Days')
plt.show()
```

System Functioning Status
This is a sanity check to ensure the data makes sense. The plot confirms that when the bike-sharing system is marked as "Non-Functioning," the rental count is zero. All the rentals come from "Functioning" days.

```python
sns.scatterplot(x='Humidity', y='Rented_Bike_Count', data=df, color='green', alpha=0.3)
plt.title('Humidity vs. Rented Bike Count')
plt.show()
```

Effect of Humidity
Here, I plotted Humidity against bike rentals. You can see a general trend where higher humidity leads to fewer rentals. People prefer riding when the air is drier and more comfortable.

```python
sns.regplot(x='Solar_Radiation', y='Rented_Bike_Count', data=df, scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Solar Radiation vs. Rented Bike Count')
plt.show()
```

Solar Radiation and Demand
This plot shows the relationship between solar radiation (sunlight intensity) and bike rentals. The red trend line goes up, indicating that people are more likely to rent bikes on sunny days with higher solar radiation.

```python
rainy_days = df[df['Rainfall'] > 0]
sns.scatterplot(x='Rainfall', y='Rented_Bike_Count', data=rainy_days)
plt.title('Rainfall vs. Rented Bike Count')
plt.show()
```

Impact of Rain
I filtered the data to look only at rainy days. The scatter plot shows a dense cluster at low rental counts when rainfall increases. Basically, even a little bit of rain significantly drops the number of people renting bikes.

```python
df['Day_of_Week'] = df['Date'].dt.day_name()
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(12, 6))
sns.boxplot(x='Day_of_Week', y='Rented_Bike_Count', data=df, order=order)
plt.title('Bike Rentals by Day of Week')
plt.show()
```

Weekly Rental Patterns
I broke down the rentals by day of the week. The usage is consistently high from Monday to Friday, which reinforces the idea that these bikes are used for commuting. There is a noticeable dip on **Sunday**, suggesting fewer people ride for leisure compared to work.

```python
sns.lineplot(x='Wind_speed', y='Rented_Bike_Count', data=df)
plt.title('Wind Speed vs. Rented Bike Count')
plt.show()
```

Wind Speed vs. Rentals
This line plot explores if wind speed matters. It looks like rentals are fairly stable with average wind, but there's a bit of fluctuation when the wind gets very strong. However, it's not as major a factor as temperature.

```python
df['Snow_Status'] = df['Snowfall'].apply(lambda x: 'Snow' if x > 0 else 'No Snow')
sns.barplot(x='Snow_Status', y='Rented_Bike_Count', data=df)
plt.title('Impact of Snowfall on Rentals')
plt.show()
```

Effect of Snow
I created a new category to compare days with snow versus days without. The bar chart makes it very clear: bike rentals drop drastically when there is snow on the ground.
